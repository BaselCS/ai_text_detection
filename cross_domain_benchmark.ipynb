```python
# Cross-Domain Benchmark  (Writing Prompts amd Essays) from MTG
import pandas as pd 
ESSAY=pd.read_csv("data/MTG_Benchmark/Essay_LLMs.csv")
WP=pd.read_csv("data/MTG_Benchmark/WP_LLMs.csv")
```

```python
ESSAY.columns
```

```python
ESSAY=ESSAY[["GPT4All","human"]]
WP=WP[["GPT4All","human"]]
```

```python
# split into train and test 80- 20
from sklearn.model_selection import train_test_split
_, ESSAY_test = train_test_split(ESSAY, test_size=0.2, random_state=999)
_, WP_test = train_test_split(WP, test_size=0.2, random_state=999)
ESSAY_test.to_csv("data/MTG_Benchmark/Cross-Domain/Essay_LLMs_test.csv", index=False)
WP_test.to_csv("data/MTG_Benchmark/Cross-Domain/WP_LLMs_test.csv", index=False)
```
